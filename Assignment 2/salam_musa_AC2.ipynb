```python
import math

def count_primes_up_to(x):
    """Count how many primes exist up to a given number x."""
    if x < 2:
        return 0
   # Initialize sieve
    sieve = [True] * (x + 1)
    sieve[0] = sieve[1] = False
    # Apply Sieve of Eratosthenes
    for p in range(2, int(math.sqrt(x)) + 1):
        if sieve[p]:  # Found a prime number
            for multiple in range(p * p, x + 1, p):
                sieve[multiple] = False  # Mark multiples as composite

   # Count primes
    return sum(sieve)  # Count how many True values remain

# Example usage
n = 122496
x = 5000000 + 122496
num_primes = count_primes_up_to(x)
print(f"Number of primes up to {x} is {num_primes}")
```
